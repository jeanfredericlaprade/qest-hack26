## Adding constraints to the Hamiltonian - an example
For demonstration purposes, let us consider a problem with 2 qubits and a simple Hamiltonian.

$$
    H = \sum_{i=0}^1 Z_i.
$$

This expression can be written in `SparsePauliOp`.

```python
from qiskit.quantum_info import SparsePauliOp

H = SparsePauliOp.from_list([("IZ", 1.0), ("ZI", 1.0)])
```

Let's consider the following equality constraint :

$$
Z_0 = 1
$$

This means that qubit 0 must be in the state $\left|0\right>$, since the average value of the observable $Z_0$ will only be equal to $1$ when the qubit is in this state.

### Transformation of the equality constraint

To incorporate this constraint into the Hamiltonian, we rewrite it as a distance expression.

We start by reformulating the previous equation

$$
Z_0 - 1 = 0, 
$$

Thus, the distance function is
$$
(Z_0 - 1)^2
$$

Let's write it as a `SparsePauliOp` object.

```python
constraint = (SparsePauliOp.from_list([("IZ", 1.0), ("II", -1.0)])) ** 2
```

This expression measures the distance between the result of the measurement of the observable $Z_0$ and the value $1$. If this distance is equal to zero, it means that the constraint $Z_0 = 1$ is satisfied. Squaring this expression keeps the distance always positive, which will be useful when adding this constraint to our cost function (Hamiltonian).

### Tolerance and penality associated to the constraint

To indicate our tolerance for non-compliance with this constraint, we introduce a parameter $\alpha$, which represents the penalty associated with breaking the constraint. This parameter $\alpha$ becomes a hyperparameter of your optimization and multiplies the distance expression. For example, for a Hamiltonian $H$, the constraint expression for the latter is obtained by 

$$
    H_c = H + \alpha(Z_0 - 1)^2.
$$

```python
alpha = 1
H_c = H + alpha * constraint
```

We're looking for an $\alpha$ such that breaking the constraint $Z_0 = 1$ is costly enough that our optimization process chooses not to do so.

It is recommended to start with a value of $\alpha$ slightly higher than the sum of the weights in the unconstrained cost function. However, too high an $\alpha$ could make the QAOA algorithm unstable, so special attention must be paid to this balance.
