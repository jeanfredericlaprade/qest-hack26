# Encoding the problem with qubits

To model this optimization problem with qubits, it is essential to encode both the observation posts and the order in which the drone must visit them. Here are some key points to consider:

- **Observation posts** : Each observation post can be represented by a qubit. At each stage of the patrol, we can indicate whether the drone is at a given post by placing the corresponding qubit in the state |1⟩. For example, if we have 4 observation posts, we will need $N=4$ qubits.

- **Route order**: Our encoding must also reflect the order in which the drone visits the observation posts. If there are 4 observation posts and each is visited exactly once, the encoding must represent the four steps of the partol. In a time-position encoding scheme, this may require $M=4$ steps and $N=4$ posts, for a total of $M*N$ qubits to encode the route.
